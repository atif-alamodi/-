```python
# ===================================================
# مشروع نظام السوبر ماركت الذكي
# مقرر البرمجة باستخدام البايثون PYT103
# الجامعة السعودية الإلكترونية - الكلية التطبيقية
# شعبة 1 - مجموعة 1
# ===================================================

# --------------------------------------------------
# المهمة 1: بناء هيكل البيانات
# الاسم: عاطف محمد العمودي
# الرقم الأكاديمي: 251004469
# --------------------------------------------------
# إنشاء قاموس المنتجات
# كل منتج هو مفتاح (key) وقيمته قاموس داخلي يحتوي على السعر (price) والكمية (qty)

products = {
    "أرز": {"price": 25.00, "qty": 50},
    "سكر": {"price": 12.50, "qty": 40},
    "حليب": {"price": 7.00, "qty": 100},
    "خبز": {"price": 3.50, "qty": 80},
    "بيض": {"price": 15.00, "qty": 60},
    "جبن": {"price": 22.00, "qty": 35},
    "دجاج": {"price": 30.00, "qty": 45},
    "ماء": {"price": 2.00, "qty": 200},
    "عصير": {"price": 8.50, "qty": 70},
    "شاي": {"price": 18.00, "qty": 55}
}

# سلة المشتريات
cart = {}


# --------------------------------------------------
# المهمة 3: نظام التحقق (Validation)
# الاسم: ريما
# --------------------------------------------------
# التحقق من وجود المنتج في النظام
def check_product(products, product_name):
    if product_name in products:
        return True
    else:
        print("المنتج غير موجود.")
        return False


# التحقق من توفر الكمية في المخزون
def check_stock(products, product_name, quantity):
    if quantity <= products[product_name]["qty"]:
        return True
    else:
        print("الكمية المطلوبة غير متوفرة.")
        print(f"المتوفر: {products[product_name]['qty']}")
        return False


# --------------------------------------------------
# المهمة 6: معالجة الأخطاء
# الاسم: فوزية
# --------------------------------------------------
# دالة للحصول على رقم صحيح من المستخدم مع معالجة الأخطاء
def get_quantity():
    while True:
        try:
            qty = int(input("أدخل الكمية المطلوبة: "))
            if qty > 0:
                return qty
            else:
                print("يجب أن تكون الكمية أكبر من صفر.")
        except ValueError:
            print("الرجاء إدخال رقم صحيح.")


# --------------------------------------------------
# المهمة 5: واجهة المخرجات (عرض المنتجات)
# الاسم: نورة
# --------------------------------------------------
# عرض جميع المنتجات المتوفرة بشكل مرتب
def display_products(products):
    print("\n" + "-" * 55)
    print(f"{'اسم المنتج':<15}{'السعر':<15}{'الكمية المتوفرة':<20}")
    print("-" * 55)

    for name, info in products.items():
        price = info["price"]
        qty = info["qty"]

        if qty == 0:
            qty_display = "غير متوفر"
        else:
            qty_display = qty

        print(f"{name:<15}{price:<15.2f}{qty_display:<20}")

    print("-" * 55)


# --------------------------------------------------
# المهمة 4: المحرك الحسابي
# الاسم: صابر
# --------------------------------------------------
# حساب المجموع الفرعي
def calculate_subtotal(products, cart):
    subtotal = 0
    for item_name, qty in cart.items():
        subtotal += products[item_name]["price"] * qty
    return subtotal


# حساب الضريبة 15%
def calculate_tax(subtotal):
    return subtotal * 0.15


# حساب الإجمالي النهائي
def calculate_final_total(subtotal, tax):
    return subtotal + tax


# تنسيق وطباعة الفاتورة النهائية
def print_invoice(products, cart, subtotal, tax, final_total):
    print("\n" + "=" * 60)
    print("                   الفاتورة النهائية")
    print("=" * 60)
    print(f"{'المنتج':<15}{'سعر الوحدة':<15}{'الكمية':<10}{'الإجمالي':<15}")
    print("-" * 60)

    for item_name, qty in cart.items():
        price = products[item_name]["price"]
        line_total = price * qty
        print(f"{item_name:<15}{price:<15.2f}{qty:<10}{line_total:<15.2f}")

    print("-" * 60)
    print(f"{'المجموع الفرعي:':<40}{subtotal:.2f} ريال")
    print(f"{'الضريبة (15%):':<40}{tax:.2f} ريال")
    print(f"{'الإجمالي النهائي:':<40}{final_total:.2f} ريال")
    print("=" * 60)


# --------------------------------------------------
# المهمة 2: إدارة عملية الشراء
# الاسم: شهد خالد والعنود
# --------------------------------------------------
# إضافة المنتج إلى السلة مع ربط التحقق ومعالجة الأخطاء
def add_to_cart(products, cart):
    product_name = input("أدخل اسم المنتج: ").strip()

    # التحقق من وجود المنتج
    if not check_product(products, product_name):
        return

    # إدخال الكمية
    qty = get_quantity()

    # التحقق من توفر الكمية
    if not check_stock(products, product_name, qty):
        return

    # خصم الكمية من المخزون
    products[product_name]["qty"] -= qty

    # إضافة المنتج إلى السلة
    if product_name in cart:
        cart[product_name] += qty
    else:
        cart[product_name] = qty

    print(f"تمت إضافة {qty} من {product_name} إلى السلة بنجاح.")


# عرض محتويات السلة
def display_cart(products, cart):
    if not cart:
        print("السلة فارغة.")
        return

    print("\n" + "-" * 55)
    print("محتويات السلة:")
    print("-" * 55)
    print(f"{'المنتج':<15}{'الكمية':<10}{'سعر الوحدة':<15}{'الإجمالي':<15}")

    for item_name, qty in cart.items():
        price = products[item_name]["price"]
        line_total = price * qty
        print(f"{item_name:<15}{qty:<10}{price:<15.2f}{line_total:<15.2f}")

    print("-" * 55)


# الدالة الرئيسية للتسوق
def shopping(products):
    cart = {}

    print("=" * 60)
    print("            مرحبًا بك في نظام السوبر ماركت الذكي")
    print("=" * 60)

    while True:
        print("\nالقائمة الرئيسية:")
        print("(1) إضافة منتج")
        print("(2) عرض السلة")
        print("(3) عرض المنتجات")
        print("(4) إنهاء التسوق والدفع")

        choice = input("اختر رقم العملية: ").strip()

        if choice == "1":
            add_to_cart(products, cart)

        elif choice == "2":
            display_cart(products, cart)

        elif choice == "3":
            display_products(products)

        elif choice == "4":
            if not cart:
                print("السلة فارغة، لا يوجد مشتريات.")
            else:
                subtotal = calculate_subtotal(products, cart)
                tax = calculate_tax(subtotal)
                final_total = calculate_final_total(subtotal, tax)
                print_invoice(products, cart, subtotal, tax, final_total)

            print("شكرًا لاستخدامك نظام السوبر ماركت الذكي.")
            break

        else:
            print("اختيار غير صحيح، حاول مرة أخرى.")


# --------------------------------------------------
# تشغيل البرنامج
# --------------------------------------------------
shopping(products)
```
